```python
def segment_customer(row):
    r = row["R_Score"]
    f = row["F_Score"]

    if r >= 4 and f >= 4:
        return "Champions"

    elif r >= 3 and f >= 4:
        return "Loyal Customers"

    elif r >= 4 and f >= 2:
        return "Potential Loyalists"

    elif r >= 3 and f >= 2:
        return "Need Attention"

    elif r >= 4 and f == 1:
        return "New Customers"

    elif r == 3 and f == 1:
        return "Promising"

    elif r <= 2 and f >= 4:
        return "At Risk"

    elif r <= 2 and f >= 2:
        return "Cannot Lose Them"

    else:
        return "Lost Customers"
```

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)

rfm.head()
```

```python
# Recency Score (lower recency = higher score)

rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    q=5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

# Frequency Score

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Monetary Score

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"],
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm.head()
```

## RFM Scoring

Each customer is assigned a score from 1 to 5 for:

- **Recency (R):** Lower recency (more recent purchase) receives a higher score.
- **Frequency (F):** Higher purchase frequency receives a higher score.
- **Monetary (M):** Higher spending receives a higher score.

These scores are combined to create an overall RFM segment.
