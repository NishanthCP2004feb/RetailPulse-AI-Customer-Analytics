```python
# ============================================================
# RetailPulse AI Customer Analytics
# Notebook 10 - Streamlit Deployment
# ============================================================

import os
from pathlib import Path

import joblib
import pandas as pd
import numpy as np

print("=" * 80)
print("RetailPulse AI Customer Analytics")
print("Notebook 10 - Streamlit Deployment")
print("=" * 80)

print("\nLibraries imported successfully.")
```

```python
# ============================================================
# Project Configuration
# ============================================================

PROJECT_ROOT = Path.cwd().parent

DATA_DIR = PROJECT_ROOT / "data"
RAW_DATA_DIR = DATA_DIR / "raw"
PROCESSED_DATA_DIR = DATA_DIR / "processed"

MODELS_DIR = PROJECT_ROOT / "models"

REPORTS_DIR = PROJECT_ROOT / "reports"

STREAMLIT_DIR = PROJECT_ROOT / "streamlit"

PAGES_DIR = STREAMLIT_DIR / "pages"

UTILS_DIR = STREAMLIT_DIR / "utils"

ASSETS_DIR = STREAMLIT_DIR / "assets"

CONFIG_DIR = STREAMLIT_DIR / ".streamlit"

print("=" * 80)
print("PROJECT CONFIGURATION")
print("=" * 80)

print(f"Project Root     : {PROJECT_ROOT}")
print(f"Processed Data   : {PROCESSED_DATA_DIR}")
print(f"Models           : {MODELS_DIR}")
print(f"Reports          : {REPORTS_DIR}")
print(f"Streamlit Folder : {STREAMLIT_DIR}")
```

```python
# ============================================================
# Create Streamlit Folder Structure
# ============================================================

folders = [
    STREAMLIT_DIR,
    PAGES_DIR,
    UTILS_DIR,
    ASSETS_DIR,
    CONFIG_DIR
]

for folder in folders:
    folder.mkdir(parents=True, exist_ok=True)

print("=" * 80)
print("STREAMLIT PROJECT STRUCTURE")
print("=" * 80)

for folder in folders:
    print(f"✓ {folder}")
```

```python
# ============================================================
# Verify Processed Datasets
# ============================================================

required_datasets = [
    "retail_cleaned.csv",
    "analysis_data.csv",
    "customer_rfm.csv",
    "customer_segments.csv",
    "customer_features.csv",
    "customer_churn_predictions.csv",
    "high_risk_customers.csv",
    "inventory_analysis.csv",
    "inventory_recommendations.csv",
    "customer_product_recommendations.csv"
]

verification = []

for file in required_datasets:

    path = PROCESSED_DATA_DIR / file

    verification.append({
        "Dataset": file,
        "Exists": path.exists()
    })

verification_df = pd.DataFrame(verification)

print("=" * 80)
print("DATASET VERIFICATION")
print("=" * 80)

display(verification_df)

print()

print(
    f"Datasets Found : {verification_df['Exists'].sum()} / {len(required_datasets)}"
)
```

```python
# ============================================================
# Verify Trained Models
# ============================================================

required_models = [
    "sales_forecasting_model.pkl",
    "customer_churn_model.pkl",
    "kmeans_model.pkl",
    "scaler.pkl"
]

verification = []

for file in required_models:

    path = MODELS_DIR / file

    verification.append({
        "Model": file,
        "Exists": path.exists()
    })

model_df = pd.DataFrame(verification)

print("=" * 80)
print("MODEL VERIFICATION")
print("=" * 80)

display(model_df)

print()

print(
    f"Models Found : {model_df['Exists'].sum()} / {len(required_models)}"
)
```

```python
# ============================================================
# Create Application Folder Structure
# ============================================================

APP_DIR = PROJECT_ROOT / "app"

PAGES_DIR = APP_DIR / "pages"

UTILS_DIR = APP_DIR / "utils"

ASSETS_DIR = APP_DIR / "assets"

CONFIG_DIR = APP_DIR / ".streamlit"

folders = [
    APP_DIR,
    PAGES_DIR,
    UTILS_DIR,
    ASSETS_DIR,
    CONFIG_DIR,
]

for folder in folders:
    folder.mkdir(parents=True, exist_ok=True)

print("=" * 80)
print("APPLICATION FOLDER STRUCTURE")
print("=" * 80)

for folder in folders:
    print("✓", folder.relative_to(PROJECT_ROOT))
```

```python
# ============================================================
# Project Asset Summary
# ============================================================

from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

DATA_DIR = PROJECT_ROOT / "data"
PROCESSED_DATA_DIR = DATA_DIR / "processed"

MODELS_DIR = PROJECT_ROOT / "models"

REPORTS_DIR = PROJECT_ROOT / "reports"

processed_files = sorted(PROCESSED_DATA_DIR.glob("*.csv"))
model_files = sorted(MODELS_DIR.glob("*.pkl"))
report_files = sorted(REPORTS_DIR.glob("*.csv"))

print("=" * 80)
print("PROJECT ASSET SUMMARY")
print("=" * 80)

print(f"Processed Datasets : {len(processed_files)}")
print(f"Models             : {len(model_files)}")
print(f"Reports            : {len(report_files)}")

print("\nProcessed Files")
for file in processed_files:
    print("✓", file.name)

print("\nModels")
for file in model_files:
    print("✓", file.name)

print("\nReports")
for file in report_files:
    print("✓", file.name)
```

```python
# ============================================================
# Load Core Assets
# ============================================================

analysis_data = pd.read_csv(PROCESSED_DATA_DIR / "analysis_data.csv")
customer_segments = pd.read_csv(PROCESSED_DATA_DIR / "customer_segments.csv")
customer_rfm = pd.read_csv(PROCESSED_DATA_DIR / "customer_rfm.csv")
daily_sales = pd.read_csv(PROCESSED_DATA_DIR / "daily_sales.csv")

forecast_model = joblib.load(MODELS_DIR / "best_forecasting_model.pkl")
kmeans_model = joblib.load(MODELS_DIR / "kmeans_model.pkl")
churn_model = joblib.load(MODELS_DIR / "customer_churn_model.pkl")

print("=" * 80)
print("CORE ASSETS LOADED")
print("=" * 80)

print("Analysis Data :", analysis_data.shape)
print("Customer Segments :", customer_segments.shape)
print("Customer RFM :", customer_rfm.shape)
print("Daily Sales :", daily_sales.shape)

print()

print("Forecast Model :", type(forecast_model).__name__)
print("KMeans Model :", type(kmeans_model).__name__)
print("Churn Model :", type(churn_model).__name__)
```

```python
# ============================================================
# Deployment Readiness Check
# ============================================================

checks = {
    "Processed Data": len(processed_files) >= 8,
    "Models": len(model_files) >= 4,
    "Reports": len(report_files) >= 5,
    "Application Folder": APP_DIR.exists(),
}

print("=" * 80)
print("DEPLOYMENT READINESS")
print("=" * 80)

for name, status in checks.items():
    print(f"{name:<25} : {'PASS' if status else 'FAIL'}")

print()

if all(checks.values()):
    print("✅ Project is ready for Streamlit application development.")
else:
    print("⚠️ Some assets are missing. Review the project before deployment.")
```
